```r
install.packages("sqldf")
library(sqldf)
```

```r
data <- read.csv("cleaned_main.csv", stringsAsFactors=FALSE)
```

#Delay By Zone

```r
sqldf("
SELECT dropoff_zone,
       COUNT(*) AS total_deliveries,
       SUM(delay_flag) AS delayed_deliveries
FROM data
GROUP BY dropoff_zone
ORDER BY delayed_deliveries DESC
")
```

# Complaint Impact

```r
sqldf("
SELECT has_complaint,
       AVG(delay_flag) AS delay_rate
FROM data
GROUP BY has_complaint
")
```

# Driver Behaviour (Root Cause)

```r
sqldf("
SELECT driver_id,
       AVG(manual_route_override_count) AS avg_overrides,
       AVG(delay_flag) AS delay_rate
FROM data
GROUP BY driver_id
ORDER BY avg_overrides DESC
")
```

# Cost Analysis

```r
sqldf("
SELECT dropoff_zone,
       AVG(fuel_or_charge_cost) AS avg_cost
FROM data
GROUP BY dropoff_zone
ORDER BY avg_cost DESC
")
```

# Combined Failure Analysis

```r
sqldf("
SELECT dropoff_zone,
       COUNT(*) AS total,
       SUM(delay_flag) AS delays,
       SUM(has_complaint) AS complaints
FROM data
GROUP BY dropoff_zone
")
```
