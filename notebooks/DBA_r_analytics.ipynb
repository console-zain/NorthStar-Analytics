```r
install.packages("sqldf")
```

```r
library(ggplot2)

data <- read.csv("cleaned_main.csv")
```

# Delays by Zone

```r
ggplot(data, aes(x=dropoff_zone, fill=as.factor(delay_flag))) +
  geom_bar(position="dodge") +
  labs(title="Delays by Zone", fill="Delayed") +
  theme(axis.text.x = element_text(angle=45))
```

# Delivery Time Distribution

```r
ggplot(data, aes(x=delivery_time_hours)) +
  geom_histogram(bins=30) +
  labs(title="Delivery Time Distribution")
```

 # Cost by Zone

```r
ggplot(data, aes(x=dropoff_zone, y=fuel_or_charge_cost)) +
  geom_boxplot() +
  labs(title="Cost Distribution by Zone") +
  theme(axis.text.x = element_text(angle=45))
```

# TREND ANALYSIS

```r
ggplot(data, aes(x=delivery_time_hours, y=fuel_or_charge_cost)) +
  geom_point(alpha=0.5) +
  labs(title="Delivery Time vs Cost")
```
